# Inspect a YCSB result
Set the path below to a saved run-phase result.json artifact.

```python
import json
from pathlib import Path

result_path = Path("path/to/result.json")
result = json.loads(result_path.read_text())
print(result["throughput_ops_s"], "operations per second")
for operation, measurements in result["operations"].items():
    print(operation, measurements["count"], measurements["p99_us"])
```
