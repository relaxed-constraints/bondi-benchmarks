```python
from empty_project.hello import hello_world

print(hello_world("World"))
```

```python
import matplotlib.pyplot as plt
import pandas as pd

# Create greetings with the reusable helper and measure their lengths.
names = pd.DataFrame({"name": ["World", "Uwe", "Thomas", "Student"]})
names["greeting"] = names["name"].map(hello_world)
names["greeting_length"] = names["greeting"].str.len()
display(names)

# pandas uses Matplotlib to draw the chart.
fig, ax = plt.subplots(figsize=(6, 4), dpi=150)
names.plot.bar(
    x="greeting",
    y="greeting_length",
    legend=False,
    rot=0,
    ax=ax,
)
ax.set_ylabel("Number of characters")
ax.set_xlabel("Greeting")
ax.set_title("Greeting lengths")
fig.tight_layout()
plt.show()
```
